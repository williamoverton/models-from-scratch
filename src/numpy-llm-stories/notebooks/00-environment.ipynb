## Load our trained tokenizer

Jupyter starts this notebook's kernel in the `notebooks/` directory. The path
below goes three directories up to the repository root. The tokenizer artifact
is local and ignored by Git; if missing, follow the
[shared tokenizer training instructions](../../../libs/model-tokenizer/README.md).

We only encode and decode text here. Embeddings, position information, and the
transformer will be explored in subsequent notebooks as we implement them.


```python
tokenizer_path = Path("../../../trained-models/tokenizers/simple-english.json").resolve()
tokenizer = Tokenizer.load(tokenizer_path)

text = "Once upon a time, there was a fluffy bunny."
token_ids = tokenizer.encode(text)
print(f"Vocabulary: {tokenizer.vocab_size:,}")
print(f"Control IDs: {dict(tokenizer.control_token_ids)}")
print(f"Token IDs: {token_ids}")
print(f"Decoded: {tokenizer.decode(token_ids)}")
```
